# Briefly explain why you chose the tool

Fastp and Trimmomatic are the standard tools used for trimming, so I went with trimmomatic for a comparison. Trimmomatic is also a bit more customizable than fastp. You get more control over what you want to trim out of your fastq files.

Regarding quality report generation, there are not a lot of tools to choose from, and it doesn't really matter at the end of the day, the stats are going to be the same no matter what tool you use, you could even do it manually from the fastq files, it's the formatting that makes fastqc the standard choice. I went with Sequali because I've never tried it, but the documentation for it was available on https://docs.seqera.io/, so I tried to see what it outputs.

```python
# Folder organization

!mkdir SequaliBeforeTrimming
!mkdir SequaliAfterTrimming
!mkdir MultiQCBeforeTrimming
!mkdir MultiQCAfterTrimming
!mkdir TrimmingTrimmomatic

# Quality report; Sequali tool

!sequali SRR360498_1.filt.fastq.gz SRR360498_2.filt.fastq.gz --outdir SequaliBeforeTrimming

# MultiQC

!multiqc SequaliBeforeTrimming -o MultiQCBeforeTrimming

# Trimming: Trimmomatic

!trimmomatic PE \
  -threads 4 \
  SRR360498_1.filt.fastq.gz \
  SRR360498_2.filt.fastq.gz \
  TrimmingTrimmomatic/SRR360498_1.trimmed.paired.fastq.gz \
  TrimmingTrimmomatic/SRR360498_1.trimmed.unpaired.fastq.gz \
  TrimmingTrimmomatic/SRR360498_2.trimmed.paired.fastq.gz \
  TrimmingTrimmomatic/SRR360498_2.trimmed.unpaired.fastq.gz \
  LEADING:3 \
  TRAILING:3 \
  SLIDINGWINDOW:4:20 \
  MINLEN:50
                    
# PostTrimming Sequali

!sequali TrimmingTrimmomatic/SRR360498_1.trimmed.paired.fastq.gz TrimmingTrimmomatic/SRR360498_2.trimmed.paired.fastq.gz --outdir SequaliAfterTrimming

# PostTrimming MultiQC

!multiqc SequaliAfterTrimming -o MultiQCAfterTrimming
```
